## 1. Configuração

Execute esta célula uma vez no início da sessão. Ela importa as bibliotecas e define símbolos e funções auxiliares usadas nos demais blocos.

```python
import sympy as sp
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, Math

sp.init_printing()

x, y, z, t = sp.symbols("x y z t", real=True)
e, pi, inf = sp.exp(1), sp.pi, sp.oo
C = sp.symbols("C")

def mostrar_resultado(rotulo, expressao):
    """Exibe uma resposta matemática identificada."""
    display(Math(r"\text{" + rotulo + r"}: " + sp.latex(expressao)))

def resolver_zeros(expressao, variavel=x):
    """Retorna soluções reais de expressão igual a zero."""
    solucoes = sp.solve(sp.Eq(expressao, 0), variavel)
    return [sp.simplify(s) for s in solucoes if s.is_real is not False]

def tabela_resultados(linhas):
    """Exibe resultados curtos em tabela."""
    display(pd.DataFrame(linhas, columns=["Resultado", "Valor"]))
```

## 2. Limites

### Calcular limite de uma função

Use para enunciados como “calcule o limite”, “determine o limite quando x tende a a”, inclusive limites laterais e limites no infinito. Altere `funcao`, `ponto` e, se necessário, `direcao`. Use `"+" ` para a direita, `"-"` para a esquerda e `"+-"` para o limite bilateral.

```python
# ENTRADAS
funcao = (x**2 - 1)/(x - 1)
ponto = 1
direcao = "+-"  # "+-" bilateral, "+" direita, "-" esquerda

# CÁLCULO E RESPOSTA
limite = sp.limit(funcao, x, ponto, dir=direcao)
mostrar_resultado("Limite", limite)
```

### Limites no infinito

Use quando a variável tende a (+\infty) ou (-\infty), especialmente para determinar comportamento assintótico.

```python
# ENTRADAS
funcao = (3*x**2 + 1)/(2*x**2 - 5)
ponto = sp.oo  # sp.oo ou -sp.oo

# RESPOSTA
limite = sp.limit(funcao, x, ponto)
mostrar_resultado("Limite no infinito", limite)
```

### Continuidade em um ponto

Use para verificar se uma função é contínua em (x=a). Para funções definidas por partes, informe a expressão com `sp.Piecewise`; o bloco compara o valor da função com os limites laterais.

```python
# ENTRADAS
funcao = sp.Piecewise((x**2, x < 1), (2*x - 1, True))
ponto = 1

# CÁLCULOS
valor = sp.simplify(funcao.subs(x, ponto))
limite_esquerdo = sp.limit(funcao, x, ponto, dir="-")
limite_direito = sp.limit(funcao, x, ponto, dir="+")

# RESPOSTA
tabela_resultados([
    ["f(a)", valor],
    ["Limite à esquerda", limite_esquerdo],
    ["Limite à direita", limite_direito],
    ["Contínua no ponto", sp.simplify(valor-limite_esquerdo)==0 and sp.simplify(valor-limite_direito)==0]
])
```

## 3. Derivadas

### Derivar uma função

Use para calcular a primeira derivada, derivadas de ordem superior ou derivadas parciais. Altere apenas `funcao`, `variavel` e `ordem`. Para derivada parcial, informe a variável em relação à qual derivar.

```python
# ENTRADAS
funcao = x**3 * sp.sin(x)
variavel = x
ordem = 1  # 1 = primeira derivada; 2 = segunda; etc.

# RESPOSTA
derivada = sp.diff(funcao, variavel, ordem)
mostrar_resultado(f"Derivada de ordem {ordem}", derivada)
```

### Derivar e avaliar em um ponto

Use quando o enunciado pedir o valor da derivada em (x=a), a inclinação da tangente ou a taxa instantânea de variação.

```python
# ENTRADAS
funcao = x**3 - 4*x
ponto = 2

# CÁLCULO E RESPOSTA
derivada = sp.diff(funcao, x)
valor_derivada = sp.simplify(derivada.subs(x, ponto))
mostrar_resultado("f'(x)", derivada)
mostrar_resultado("f'(a)", valor_derivada)
```

### Derivação implícita

Use quando (x) e (y) aparecem misturados em uma relação, como (F(x,y)=0), e o exercício pedir (dy/dx). Informe a relação e, opcionalmente, um ponto para avaliar a inclinação.

```python
# ENTRADAS
relacao = x**2 + y**2 - 25  # relação escrita como F(x,y) = 0
ponto = {x: 3, y: 4}         # use None para não avaliar em um ponto

# CÁLCULO
derivada_implicita = sp.factor(-sp.diff(relacao, x) / sp.diff(relacao, y))
mostrar_resultado("dy/dx", derivada_implicita)

# RESPOSTA OPCIONAL NO PONTO
if ponto is not None:
    mostrar_resultado("Inclinação no ponto", sp.simplify(derivada_implicita.subs(ponto)))
```

### Reta tangente e reta normal

Use quando o enunciado pedir a equação da reta tangente ou normal ao gráfico (y=f(x)) em (x=a).

```python
# ENTRADAS
funcao = x**2 + 1
ponto_x = 2

# CÁLCULO
ponto_y = sp.simplify(funcao.subs(x, ponto_x))
inclinacao = sp.simplify(sp.diff(funcao, x).subs(x, ponto_x))
reta_tangente = sp.expand(ponto_y + inclinacao*(x-ponto_x))

# RESPOSTAS
mostrar_resultado("Ponto de tangência", sp.Tuple(ponto_x, ponto_y))
mostrar_resultado("Reta tangente", sp.Eq(y, reta_tangente))
if inclinacao == 0:
    mostrar_resultado("Reta normal", sp.Eq(x, ponto_x))
else:
    reta_normal = sp.expand(ponto_y - (x-ponto_x)/inclinacao)
    mostrar_resultado("Reta normal", sp.Eq(y, reta_normal))
```

### Resolver uma equação ou encontrar raízes

Use para encontrar zeros, interceptos no eixo (x), pontos em que (f(x)=k) ou soluções de uma equação. Altere `equacao` e `variavel`.

```python
# ENTRADAS
equacao = x**3 - 6*x**2 + 11*x - 6  # expressão igualada a zero
variavel = x

# RESPOSTA
solucoes = sp.solve(sp.Eq(equacao, 0), variavel)
display(pd.DataFrame({"Soluções": solucoes}))
```

## 4. Aplicações de derivadas

### Pontos críticos e extremos locais

Use para encontrar pontos estacionários, isto é, onde (f'(x)=0), e candidatos a máximos ou mínimos locais. O bloco lista os pontos e classifica-os pelo teste da segunda derivada quando possível. Pontos onde a derivada não existe exigem análise adicional.

```python
# ENTRADAS
funcao = x**3 - 3*x**2 + 1

# CÁLCULO
primeira = sp.diff(funcao, x)
segunda = sp.diff(funcao, x, 2)
pontos = resolver_zeros(primeira)

# RESPOSTAS
linhas = []
for ponto in pontos:
    valor = sp.simplify(funcao.subs(x, ponto))
    concavidade = sp.simplify(segunda.subs(x, ponto))
    if concavidade > 0:
        classificacao = "Mínimo local"
    elif concavidade < 0:
        classificacao = "Máximo local"
    else:
        classificacao = "Teste inconclusivo"
    linhas.append([f"x = {ponto}; f(x) = {valor}", classificacao])

display(pd.DataFrame(linhas, columns=["Ponto crítico", "Classificação"]))
```

### Crescimento e decrescimento em um intervalo

Use quando o enunciado pedir os intervalos em que a função cresce ou decresce. Informe a função e os limites do intervalo. O bloco encontra os pontos críticos internos e testa o sinal da derivada em cada subintervalo.

```python
# ENTRADAS
funcao = x**3 - 3*x
inicio = -3
fim = 3

# CÁLCULO
derivada = sp.diff(funcao, x)
criticos = sorted([p for p in resolver_zeros(derivada) if p.is_real and inicio < p < fim], key=lambda p: float(p))
limites = [sp.sympify(inicio)] + criticos + [sp.sympify(fim)]

# RESPOSTA
linhas = []
for esquerda, direita in zip(limites[:-1], limites[1:]):
    amostra = sp.simplify((esquerda + direita)/2)
    sinal = sp.sign(derivada.subs(x, amostra))
    classificacao = "Crescente" if sinal > 0 else "Decrescente" if sinal < 0 else "Sinal inconclusivo"
    linhas.append([f"({esquerda}, {direita})", classificacao])
display(pd.DataFrame(linhas, columns=["Intervalo", "Comportamento"]))
```

### Concavidade e pontos de inflexão

Use quando for necessário determinar onde a função é côncava para cima ou para baixo e localizar candidatos a pontos de inflexão. O teste de mudança de sinal da segunda derivada deve ser confirmado nos pontos candidatos.

```python
# ENTRADAS
funcao = x**4 - 4*x**2
inicio = -3
fim = 3

# CÁLCULO
segunda = sp.diff(funcao, x, 2)
candidatos = sorted([p for p in resolver_zeros(segunda) if p.is_real and inicio < p < fim], key=lambda p: float(p))
limites = [sp.sympify(inicio)] + candidatos + [sp.sympify(fim)]

# RESPOSTA — sinal da segunda derivada em cada intervalo
linhas = []
for esquerda, direita in zip(limites[:-1], limites[1:]):
    amostra = sp.simplify((esquerda + direita)/2)
    sinal = sp.sign(segunda.subs(x, amostra))
    classificacao = "Côncava para cima" if sinal > 0 else "Côncava para baixo" if sinal < 0 else "Inconclusivo"
    linhas.append([f"({esquerda}, {direita})", classificacao])
display(pd.DataFrame(linhas, columns=["Intervalo", "Concavidade"]))

# CANDIDATOS — conferir se há mudança de concavidade
display(pd.DataFrame({"Candidatos a inflexão": candidatos}))
```

### Máximo e mínimo absolutos em intervalo fechado

Use em problemas que pedem o maior ou menor valor de uma função em ([a,b]). O cálculo compara os valores nos extremos do intervalo e nos pontos críticos internos.

```python
# ENTRADAS
funcao = x**3 - 3*x
inicio = -2
fim = 3

# CÁLCULO
derivada = sp.diff(funcao, x)
criticos = [p for p in resolver_zeros(derivada) if p.is_real and inicio < p < fim]
candidatos = [sp.sympify(inicio), *criticos, sp.sympify(fim)]
valores = [(p, sp.simplify(funcao.subs(x, p))) for p in candidatos]

# RESPOSTAS
ponto_minimo, valor_minimo = min(valores, key=lambda par: float(sp.N(par[1])))
ponto_maximo, valor_maximo = max(valores, key=lambda par: float(sp.N(par[1])))
tabela_resultados([
    ["Mínimo absoluto", f"f({ponto_minimo}) = {valor_minimo}"],
    ["Máximo absoluto", f"f({ponto_maximo}) = {valor_maximo}"]
])
```

### Movimento: posição, velocidade e aceleração

Use quando a posição de um objeto for dada em função do tempo e o exercício pedir velocidade, aceleração, deslocamento ou valores em um instante. A velocidade é a derivada da posição e a aceleração é a derivada da velocidade.

```python
# ENTRADAS
posicao = t**3 - 6*t**2 + 9*t
instante = 2

# CÁLCULOS
velocidade = sp.diff(posicao, t)
aceleracao = sp.diff(velocidade, t)

# RESPOSTAS NO INSTANTE INFORMADO
tabela_resultados([
    ["Posição", sp.simplify(posicao.subs(t, instante))],
    ["Velocidade", sp.simplify(velocidade.subs(t, instante))],
    ["Aceleração", sp.simplify(aceleracao.subs(t, instante))]
])
```

### Taxas relacionadas

Use quando duas ou mais grandezas variam com o tempo e estão ligadas por uma equação — por exemplo, raio e volume, distância e ângulo. Informe a relação entre as grandezas, os valores no instante e a taxa conhecida. O bloco calcula a taxa desconhecida por diferenciação implícita em relação ao tempo.

```python
# ENTRADAS
relacao = x**2 + y**2 - 25  # relação entre as grandezas x e y
valor_x = 3
valor_y = 4
taxa_x = 2  # dx/dt

# CÁLCULO
taxa_y = sp.solve(
    sp.diff(relacao, x)*taxa_x + sp.diff(relacao, y)*sp.Symbol("taxa_y"),
    sp.Symbol("taxa_y")
)[0]
taxa_y = sp.simplify(taxa_y.subs({x: valor_x, y: valor_y}))

# RESPOSTA
mostrar_resultado("Taxa dy/dt", taxa_y)
```

Use quando o enunciado pedir aproximação de uma função perto de (x=a), polinômio de Taylor ou polinômio de Maclaurin (caso (a=0)). Informe o grau desejado.

```python
# ENTRADAS
funcao = sp.exp(x)
ponto = 0
grau = 4

# RESPOSTA
polinomio = sp.series(funcao, x, ponto, grau+1).removeO()
mostrar_resultado("Polinômio de Taylor", sp.expand(polinomio))
```

## 5. Integrais

### Calcular integral indefinida (primitiva)

Use para “calcule a integral”, “encontre uma primitiva” ou “integre a função”. Altere apenas o integrando e a variável. A constante de integração é apresentada separadamente.

```python
# ENTRADAS
integrando = 3*x**2 + 2*x - 5
variavel = x

# RESPOSTA
primitiva = sp.integrate(integrando, variavel)
mostrar_resultado("Integral indefinida", sp.Eq(sp.Symbol("F"), primitiva + C))
```

### Calcular integral definida

Use quando os limites inferior e superior forem informados. O resultado é a integral com sinal; se o exercício pedir área geométrica, use o bloco de área.

```python
# ENTRADAS
integrando = x**2
limite_inferior = 0
limite_superior = 3

# RESPOSTA
integral_definida = sp.integrate(integrando, (x, limite_inferior, limite_superior))
mostrar_resultado("Integral definida", integral_definida)
```

### Integral com substituição, por partes ou frações parciais

Na maioria dos exercícios, o SymPy escolhe automaticamente uma técnica de integração. Use o bloco de integral indefinida como primeira tentativa. Se precisar estudar a técnica utilizada, aplique `sp.integrate` a uma expressão transformada ou use os comandos específicos abaixo para obter a decomposição algébrica.

```python
# ENTRADAS
integrando = x*sp.exp(x)

# DECOMPOSIÇÕES ÚTEIS
mostrar_resultado("Decomposição em frações parciais", sp.apart(1/(x**2-1), x))
mostrar_resultado("Forma expandida", sp.expand(integrando))
mostrar_resultado("Forma fatorada", sp.factor(x**2-1))
```

### Integral imprópria

Use quando um limite de integração for infinito ou quando o integrando apresentar uma singularidade no intervalo. Informe a integral e o SymPy tentará calcular o limite impróprio.

```python
# ENTRADAS
integrando = sp.exp(-x)
limite_inferior = 0
limite_superior = sp.oo

# RESPOSTA
resultado = sp.integrate(integrando, (x, limite_inferior, limite_superior))
mostrar_resultado("Integral imprópria", resultado)
```

## 6. Aplicações de integrais

### Área geométrica entre uma curva e o eixo x

Use quando o enunciado pedir a área da região entre (y=f(x)) e o eixo (x). A área geométrica soma valores positivos, mesmo quando a função fica abaixo do eixo. Informe o intervalo; para funções que cruzam o eixo, o valor absoluto trata a mudança de sinal.

```python
# ENTRADAS
funcao = x**2 - 1
inicio = -2
fim = 2

# RESPOSTA
area = sp.integrate(sp.Abs(funcao), (x, inicio, fim))
mostrar_resultado("Área geométrica", area)
```

### Área entre duas curvas

Use quando o enunciado pedir a área limitada por duas funções. Informe a curva superior e a inferior no intervalo considerado. Se elas se cruzarem dentro do intervalo, divida a região nos pontos de interseção ou use o bloco com valor absoluto.

```python
# ENTRADAS
curva_1 = x
curva_2 = x**2
inicio = 0
fim = 1

# RESPOSTA
area = sp.integrate(sp.Abs(curva_1-curva_2), (x, inicio, fim))
mostrar_resultado("Área entre as curvas", area)
```

### Volume de revolução — discos e arruelas

Use para calcular volume ao girar uma região em torno do eixo (x), quando as seções perpendiculares ao eixo são discos ou arruelas. Informe os raios externo e interno como funções de (x).

```python
# ENTRADAS
raio_externo = 2
raio_interno = x
inicio = 0
fim = 2

# RESPOSTA
volume = sp.pi*sp.integrate(raio_externo**2-raio_interno**2, (x, inicio, fim))
mostrar_resultado("Volume de revolução", volume)
```

### Volume pelo metodo das cascas cilíndricas

Use quando a região for girada em torno de um eixo vertical e a integração em relação a (x) for conveniente. Informe o raio de cada casca e a altura da região.

```python
# ENTRADAS
raio = x
altura = 4 - x**2
inicio = 0
fim = 2

# RESPOSTA
volume = 2*sp.pi*sp.integrate(raio*altura, (x, inicio, fim))
mostrar_resultado("Volume pelo método das cascas", volume)
```

### Valor médio de uma função em um intervalo

Use quando o exercício pedir o valor médio de uma função em ([a,b]).

```python
# ENTRADAS
funcao = x**2
inicio = 0
fim = 3

# RESPOSTA
valor_medio = sp.integrate(funcao, (x, inicio, fim))/(fim-inicio)
mostrar_resultado("Valor médio da função", sp.simplify(valor_medio))
```

### Comprimento de arco

Use para calcular o comprimento do gráfico (y=f(x)) em um intervalo. O integrando depende da derivada da função.

```python
# ENTRADAS
funcao = x**2
inicio = 0
fim = 1

# RESPOSTA
comprimento = sp.integrate(sp.sqrt(1+sp.diff(funcao,x)**2), (x, inicio, fim))
mostrar_resultado("Comprimento de arco", comprimento)
```

### Integração numérica

Use quando a integral não possuir forma elementar simples ou quando o enunciado pedir uma aproximação numérica. Altere integrando e limites; `precisao` controla os dígitos apresentados.

```python
# ENTRADAS
integrando = sp.exp(-x**2)
inicio = 0
fim = 1
precisao = 8

# RESPOSTA
aproximacao = sp.N(sp.Integral(integrando, (x, inicio, fim)), precisao)
mostrar_resultado("Aproximação numérica", aproximacao)
```

## 7. Análise completa de uma função

### Resumo de propriedades e gráfico

Use quando o exercício pedir um estudo mais amplo da função, como domínio, interceptos, derivadas, pontos críticos, extremos, concavidade e gráfico. Este bloco reúne informações que normalmente seriam obtidas em vários cálculos separados. Informe o intervalo de visualização; a análise algébrica de pontos críticos é feita separadamente.

```python
# ENTRADAS
funcao = x**3 - 3*x
intervalo = (-3, 3)

# CÁLCULOS ALGÉBRICOS
primeira = sp.diff(funcao, x)
segunda = sp.diff(funcao, x, 2)
zeros = resolver_zeros(funcao)
criticos = resolver_zeros(primeira)
intercepto_y = sp.simplify(funcao.subs(x, 0))

linhas = [
    ["Função", funcao],
    ["Domínio real", sp.calculus.util.continuous_domain(funcao, x, sp.S.Reals)],
    ["Intercepto no eixo y", intercepto_y],
    ["Zeros reais", zeros],
    ["Primeira derivada", primeira],
    ["Segunda derivada", segunda],
    ["Pontos críticos (x)", criticos],
    ["Valores nos pontos críticos", [sp.simplify(funcao.subs(x,p)) for p in criticos]]
]
display(pd.DataFrame(linhas, columns=["Propriedade", "Resultado"]))

# GRÁFICO
funcao_numerica = sp.lambdify(x, funcao, "numpy")
valores_x = np.linspace(float(intervalo[0]), float(intervalo[1]), 500)
valores_y = funcao_numerica(valores_x)

plt.figure(figsize=(9,5))
plt.plot(valores_x, valores_y, label="f(x)")
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)
if criticos:
    pontos_y = [float(sp.N(funcao.subs(x,p))) for p in criticos if p.is_real]
    pontos_x = [float(sp.N(p)) for p in criticos if p.is_real]
    plt.scatter(pontos_x, pontos_y, label="Pontos críticos")
plt.xlim(intervalo)
plt.grid(True)
plt.legend()
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Gráfico da função")
plt.show()
```

## Observações de uso

- Expressões algébricas devem ser escritas com SymPy: use `sp.sin(x)`, `sp.cos(x)`, `sp.exp(x)`, `sp.log(x)` e `sp.sqrt(x)`.
- Potências usam `**`; multiplicações devem ser explícitas, como `2*x`.
- Para funções definidas por partes, use `sp.Piecewise((expressao_1, condicao_1), (expressao_2, True))`.
- Resultados simbólicos podem exigir condições de domínio e hipóteses adicionais. Verifique se o intervalo informado pertence ao domínio.
- Nos problemas de otimização, áreas e volumes, a modelagem da situação deve ser feita antes do cálculo simbólico.
